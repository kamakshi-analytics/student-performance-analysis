Analyze study hours and attendance

```python
print("avg of study hours:",df["Study_Hours"].mean())
print("avg of attendance:",df["Attendance"].mean())
print("avg of previous score:",df["Previous_Score"].mean())
print("avg of final score:",df["Final_Score"].mean())
```

Duplicate records and missing value

```python
print("missing values:",df.isnull().sum())
print("Duplicate values:",df.duplicated().sum())
print("unique student_ids:",df["Student_ID"].nunique())
```

Relationship between varaibles

```python
study_correlation=df["Study_Hours"].corr(df["Final_Score"])
attendance_correlation=df["Attendance"].corr(df["Final_Score"])

print("study corr final score:",round(study_correlation,3))  #strong positive correlation where corr +1
print("Attendace corr final score:",round(attendance_correlation,3))
```

Scatter Plot

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df,
               x="Study_Hours",
               y="Final_Score")

plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.grid(True,alpha=0.3)
plt.show()
```

Regression plot

```python
plt.figure(figsize=(8,5))

sns.regplot(data=df,
               x="Study_Hours",
               y="Final_Score")

plt.title("Relationship between study hours and final score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.show()
```

Compare avg scores by study hours

```python
average_scores=(df.groupby("Study_Hours")["Final_Score"].mean().reset_index())
average_scores
```

Distribution of Final scores

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df,x="Final_Score",bins=6,kde=True)

plt.title("Distributon of Final Scores")
plt.xlabel("Final Score")
plt.ylabel("Number of Students")
plt.show()
```

Compare previous and final  scores

```python
compare_score=df[["Previous_Score","Final_Score"]].mean()

plt.bar(compare_score.index,compare_score.values)
plt.title("Average previous score vs Average final score")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.show()
```

Visualize the Correlation Matrix

```python
correlation_matrix=df[
    ["Study_Hours","Attendance","Previous_Score","Final_Score"]
    ].corr()

plt.figure(figsize=(8,6))
sns.heatmap(               #Heatap  useful for correlation b/w numerical columns in onne chart
    correlation_matrix,
    annot=True,            #annotation is True
    cmap="coolwarm",       #cmap is color map
    fmt=".2f",             #format
    vmin=-1,               #min color is -1
    vmax=1)                #max color is 1
        
plt.title("Correlation matrix of student performance")
plt.show()
```


## Key Findings

1. The average final score in the sample is approximately 68.87.
2. Study hours and final scores show a strong positive correlation in this illustrative dataset.
3. Attendance and final scores also show a strong positive correlation.
4. The histogram helps visualize the distribution of final scores.
5. The correlation matrix summarizes relationships between study hours, attendance, previous scores, and final scores.

## Limitations

- The dataset contains only 15 illustrative student records.
- The data is created for practice and does not represent a real student population.
- Correlation does not imply causation.
- Larger, real-world datasets would be needed to draw meaningful general conclusions.
